```python
#
from langchain_chroma import Chroma

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os 
from dotenv import load_dotenv 

load_dotenv() 

hf_token = os.getenv('HF_TOKEN')
if hf_token is None:
    raise ValueError("HF_TOKEN is not set in the environment or .env file.")

os.environ['HF_TOKEN'] = hf_token
from langchain_huggingface import HuggingFaceEmbeddings 
```

```python
from langchain_chroma import Chroma
```

```python
loader= TextLoader('speech.txt')
document= loader.load()
document
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=20)
splits=text_splitter.split_documents(document)
```

```python
embedding=HuggingFaceEmbeddings(model_name='all-MiniLM-L6-v2')
vector_db=Chroma.from_documents(splits,embedding)
```

```python
vector_db.similarity_search_with_relevance_scores('specific languages are you interested')
```

```python
vector_db=Chroma.from_documents(documents=splits,embedding=embedding,persist_directory='./croma_db')
```
