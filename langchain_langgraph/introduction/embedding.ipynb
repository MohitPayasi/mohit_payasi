### FAISS Vector Store Initialization
- **Purpose**: LangChain documents aur HuggingFace embeddings model ka use karke ek local in-memory FAISS vector database create karna.
- **Method**: `FAISS.from_documents(documents, embedding)`
- **Parameters**:
  - `documents`: LangChain `Document` objects ki list jisme text aur metadata hota hai.
  - `embedding`: Vector embeddings generate karne wala model object (`HuggingFaceEmbeddings`).
- **Returns**:
  - `db`: Initialized FAISS vector store object, jisme similarity search query run ki ja sakti hai.
- **Important Note**: Is method mein pre-calculated embeddings pass nahi karni hoti; model khud internally embeddings compute kar leta hai.

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
url = "https://www.hindilyrics4u.com/song/choree_choree_chupke_se.htm"

loader = WebBaseLoader(
    web_path=url
)

docs = loader.load()
print(docs)
#print(docs[0].page_content)
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=10,separators='\n\n')
text_document=text_splitter.split_documents(docs)
texts = [doc.page_content for doc in text_document]
# Ab embed_documents call karein
doc_embeddings = embedding.embed_documents(texts)
doc_embeddings
```

```python
from langchain_community.vectorstores import FAISS
db = FAISS.from_texts(texts, embedding)
```

```python
db.similarity_search('chori chupke se aake meri aankho')
```

```python
query='chori chupke se aake meri aankho'
```

Retriever ek aisi interface hai jo ek sawal (query string) leti hai aur usse related sabse relevant documents (list[Document]) dhoondh kar wapas laati hai.

```python
retriever= db.as_retriever()
docs=retriever.invoke(query)
```

```python
# similarity search with score





docs_= db.similarity_search(query)
docs_
```

```python
docs_and_score= db.similarity_search_with_score(query)
docs_and_score
```
