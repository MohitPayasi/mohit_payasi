```python
from langchain_community.document_loaders import WebBaseLoader

url = "https://chatgpt.com/c/6adc8cff-1d94-83ee-b7e1-0de4aeea8474"

loader = WebBaseLoader(
    web_path=url
)

docs = loader.load()

print(docs[0].page_content)
```

```python
from bs4 import BeautifulSoup

html = docs[0].page_content

soup = BeautifulSoup(html, "html.parser")

text = soup.get_text(
    separator="\n",
    strip=True
)

print(text)
```

```python
import wikipedia
from langchain_community.document_loaders import WikipediaLoader

# 1. Set a custom User-Agent as required by Wikimedia policy
wikipedia.set_user_agent("MyLangChainApp/1.0 (contact@example.com)")

# 2. Patch set_lang to ensure HTTPS is used (the library defaults to http://)
_orig_set_lang = wikipedia.set_lang
def _set_lang_https(prefix):
    _orig_set_lang(prefix)
    wikipedia.API_URL = f"https://{prefix.lower()}.wikipedia.org/w/api.php"

wikipedia.set_lang = _set_lang_https
wikipedia.API_URL = "https://en.wikipedia.org/w/api.php"

# 3. Load documents
docs = WikipediaLoader(query="narendra modi", load_max_docs=3).load()
docs
```

```python
# HTML Text splitter
```

```python
from langchain_text_splitters import HTMLHeaderTextSplitter
```

```python
html_string = """
<html>
    <body>
        <h1>Hello World</h1>
        <p>Welcome to Python</p>
    </body>
</html>
"""

print(html_string)
```

```python

header_to_split_on=[
('h1','Header 1'),
('h2','Header 2')

]
html_splitter=HTMLHeaderTextSplitter(header_to_split_on)
```

```python
html_splitter.split_text(html_string)
```
