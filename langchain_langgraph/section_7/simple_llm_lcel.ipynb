LCEL ka full form hai LangChain Expression Language.

Yeh LangChain ka ek modern aur declarative tareeka hai jisse aap alag-alag components (jaise Prompt, Model/LLM, aur Output Parser) ko aapas mein aasaani se jodh kar ek pipeline (chain) bana sakte hain.

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['GROQ_API_KEY']=os.getenv('GROQ_API_KEY')
os.environ['GROQ_API_KEY']= os.getenv('GROQ_API_KEY')
os.environ['LANGCHAIN_API_KEY']= os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_TRACING_V2']='true'
os.environ['LANGCHAIN_PROJECT']= os.getenv('LANGCHAIN_PROJECT')
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model='openai/gpt-oss-20b')
print(llm)
```

```python
from langchain_core.messages import HumanMessage,SystemMessage,AIMessage
```

```python
msg= [
SystemMessage(content='translate english into hindi '),
HumanMessage(content='hello how are you')
]
```

```python
responce=llm.invoke(msg)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
str= StrOutputParser()
```

```python
str.invoke(responce)
```

```python
# using LCEL we can create chain
chain= llm|str
chain.invoke(msg)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
generic_template= 'translate the following into {langauge}'
prompt= ChatPromptTemplate.from_messages([
    ('system', generic_template),('user',"{text}")
])
```

```python
result=prompt.invoke({'langauge':'hindi','text':'hello how are you ' })
result.to_messages()
```

```python
chain=prompt|llm|str
chain.invoke({'langauge':'hindi','text':'hello how are you ' })
```
