```python
docs= loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter= RecursiveCharacterTextSplitter(chunk_size=2000,
keep_separator=True,is_separator_regex=True,chunk_overlap=300)
documents=splitter.split_documents(docs)
```

### `create_stuff_documents_chain` - Combine Documents Chain

- **Purpose**: Ye chain retrieved documents ko "stuff" (concatenate) karke prompt ke `{context}` placeholder mein inject karti hai aur LLM ko pass karti hai taaki context ke basis par answer generate ho sake.
- **Parameters**:
  - `llm`: Language model instance (jaise aapka `llm = ChatGroq(...)`).
  - `prompt`: `ChatPromptTemplate` jisme `{context}` aur `{input}` variables define hon.
  - `document_variable_name` *(optional)*: Prompt template ke context variable ka naam (default: `'context'`).
- **Returns**: Ek runnable chain jo `{'context': [...], 'input': '...'}` lekar formatted answer return karti hai.
- **Logic**: Saare documents ek hi prompt mein pack hokar single LLM call mein process hote hain (simple and fast).


```python
from langchain_classic.chains.combine_documents import (
    create_stuff_documents_chain

)
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt= ChatPromptTemplate.from_template(
""" answer the following question based on provided context

<context>
{context}
</context>
"""
)
```

```python
from langchain_core.documents import Document
```

```python
document_chain=create_stuff_documents_chain(llm, prompt)

document_chain.invoke({'input':'langsmith has two uses limits : total trances and extended',
'context':[Document(page_content="langsmith has two uses limits : total trances and extended")]})
```
