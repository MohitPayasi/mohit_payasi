```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_huggingface  import HuggingFaceEmbeddings
from langchain_text_splitters import CharacterTextSplitter

import os
from dotenv import load_dotenv
load_dotenv()
loader= TextLoader('speech.txt')
document=loader.load()
os.environ['HF_TOKEN']== os.getenv('HF_TOKEN')
```

```python
text_splitter= CharacterTextSplitter(chunk_size=100, chunk_overlap=30)
```

```python
docs=text_splitter.split_documents(document)
docs
```

```python
embedding= HuggingFaceEmbeddings(model_name='all-MiniLM-L6-v2')
db= FAISS.from_documents(docs,embedding)
```

```python
db._embed_query('hello')
```

```python
query='Hello is a common word used to'
db.similarity_search(query)
```

```python
retriever=db.as_retriever()
docs=retriever.invoke(query)
```

```python
db.similarity_search_with_score(query) # find manhattern distance
```

```python
embedding_vector=embedding.embed_query(query)
print(embedding_vector)
```

```python
db.similarity_search_by_vector(embedding_vector)
```
