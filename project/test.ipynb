```python
import os
from dotenv import load_dotenv
load_dotenv()
import langchain_anthropic
import langchain_groq
groq_api=os.getenv('GROQ_API')
from langchain_groq import ChatGroq
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter,CharacterTextSplitter
from langchain_community.vectorstores import FAISS,Chroma
model = ChatGroq(model="openai/gpt-oss-20b",api_key=os.getenv("groq_api"))
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

massage=[
    SystemMessage(content='Convert english to hindi'),
    HumanMessage(content=' wanna  you')
    ]
```

```python


response = model.invoke(massage)
print(response.content)
```

```python
# string output parser
from langchain_core.output_parsers import StrOutputParser
parser= StrOutputParser()
parser.invoke(response)

chain=model|parser
massage=[SystemMessage(content="give info only politics"),HumanMessage(content=input('ask anything '))]
chain.invoke(massage)
```
