```python
import os
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()

# Map custom keys to LangChain's expected standard environment variable names
if os.getenv('GROQ_API') and not os.getenv('GROQ_API_KEY'):
    os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API')
if os.getenv('google_api_key') and not os.getenv('GOOGLE_API_KEY'):
    os.environ['GOOGLE_API_KEY'] = os.getenv('google_api_key')
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage

# Initialize Groq Model
model = ChatGroq(model='openai/gpt-oss-20b')
response = model.invoke([HumanMessage(content='Hello! Explain in one sentence what you are.')])
print(response.content)
```
