```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

data = np.array([-20,10, 12, 15, 18, 20, 22, 25, 28, 30, 100])

z_scores = zscore(data)
plt.figure(figsize=(3,3))
plt.scatter(range(len(data)), z_scores)

plt.axhline(3, linestyle="--")
plt.axhline(-3, linestyle="--")

plt.xlabel("Data Point")
plt.ylabel("Z-Score")
plt.title("Z-Score Outlier Visualization")

plt.show()
```

```python
import seaborn as sns
import pandas as pd
df=sns.load_dataset('tips')
```

```python
df.shape
```

```python
plt.boxplot(df['total_bill'])
```

```python
q1= df['total_bill'].quantile(0.25)
q3= df['total_bill'].quantile(0.75)
iqr=q3-q1

lower= q1-1.5*iqr

upper= q3+1.5*iqr

df=df[(df['total_bill']<upper) & (df['total_bill']>lower)]
```

```python
plt.boxplot(df['total_bill'])
```

```python
df.shape
```

```python
plt.boxplot(df['tip'])
```

```python
q1= df['tip'].quantile(0.25)
q3= df['tip'].quantile(0.75)
iqr=q3-q1
print('iqr------------', iqr)
lower= q1-1.5*iqr
print('lower---------', lower)
upper= q3+1.5*iqr
print('upper----------', upper)
```

```python
df['tip']=df['tip'].clip(upper=upper)
plt.boxplot(df['tip'])
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder,LabelEncoder,MinMaxScaler
from sklearn.compose import ColumnTransformer
```

```python
df.columns
```

```python
num_fea=['total_bill']
ohe_fea=['sex', 'smoker']
min_max= ['tip']
```

```python
transformer=ColumnTransformer(
transformers=[('num', StandardScaler(),num_fea),
('min_max', MinMaxScaler(),min_max),
('cat', OneHotEncoder(drop='first'), ohe_fea),
], remainder='passthrough'
)
```

```python
transformer.fit_transform(df)
```

```python
pd.DataFrame(transformer.fit_transform(df), 
columns=transformer.get_feature_names_out())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

Pipeline([
    ('preprocesser',transformer),
    ('logistic_model', LogisticRegression())
])
```
